```python

print("\nOccupation Type Frequency:")
print(app['OCCUPATION_TYPE'].value_counts())

# Count Plot for Occupation Type
colors = ['red', 'blue', 'green', 'orange', 'purple',
          'pink', 'cyan', 'yellow', 'brown', 'gray']

plt.figure(figsize=(12,6))
sns.countplot(
    data=app,
    x='OCCUPATION_TYPE',
    order=app['OCCUPATION_TYPE'].value_counts().index,
    palette=colors
)
plt.xticks(rotation=90)
plt.title("Distribution of Occupation Type")
plt.xlabel("Occupation Type")
plt.ylabel("Count")
plt.show()

# Frequency Count for Income Type

print("\nIncome Type Frequency:")
print(app['NAME_INCOME_TYPE'].value_counts())

plt.figure(figsize=(8,5))
sns.countplot(
    data=app,
    x='NAME_INCOME_TYPE',
    order=app['NAME_INCOME_TYPE'].value_counts().index,
    palette=colors
)
plt.xticks(rotation=45)
plt.title("Income Type Distribution")
plt.show()

# Frequency Count for Education Type 

print("\nEducation Type Frequency:")
print(app['NAME_EDUCATION_TYPE'].value_counts())

plt.figure(figsize=(8,5))
sns.countplot(
    data=app,
    x='NAME_EDUCATION_TYPE',
    order=app['NAME_EDUCATION_TYPE'].value_counts().index,
    palette=colors
    )
plt.xticks(rotation=30)
plt.title("Education Level Distribution")
plt.show()

# Frequency Count for Family Status

print("\nFamily Status Frequency:")
print(app['NAME_FAMILY_STATUS'].value_counts())

plt.figure(figsize=(8,5))
sns.countplot(
    data=app,
    x='NAME_FAMILY_STATUS',
    order=app['NAME_FAMILY_STATUS'].value_counts().index,
    palette=colors
)
plt.xticks(rotation=20)
plt.title("Family Status Distribution")
plt.show()

# Numerical Feature - Income

plt.figure(figsize=(8,5))
sns.histplot(app['AMT_INCOME_TOTAL'], bins=30, kde=True,color='green')
plt.title("Income Distribution")
plt.xlabel("Annual Income")
plt.ylabel("Frequency")
plt.show()


# Box Plot for Income (Outlier Detection)

plt.figure(figsize=(8,5))
sns.boxplot(x=app['AMT_INCOME_TOTAL'],color='red')
plt.title("Income Box Plot")
plt.show()


# Numerical Feature - Age
# (DAYS_BIRTH is stored as negative days)

app['AGE'] = (-app['DAYS_BIRTH'] / 365).astype(int)

plt.figure(figsize=(8,5))
sns.histplot(app['AGE'], bins=25, kde=True,color='blue')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.show()

# Gender Distribution

print("\nGender Distribution:")
print(app['CODE_GENDER'].value_counts())

plt.figure(figsize=(6,4))
sns.countplot(data=app, x='CODE_GENDER')
plt.title("Gender Distribution")

plt.show()
```
