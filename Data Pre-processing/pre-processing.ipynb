```python
app=pd.read_csv('D:\craditcard approval pradiction\data coluction\Application_record.csv')
credit= pd.read_csv('D:\craditcard approval pradiction\data coluction\credit_record.csv')
app.head()
credit.head()
```

```python

# Remove duplicate rows
app = app.drop_duplicates()

print("Duplicates Removed Successfully")
```

```python
# Check missing values
print(app.isnull().sum())

# Fill missing values with the mode
app = app.fillna(app.mode().iloc[0])

print("Missing Values Handled")
```

```python

# Merge datasets using ID
data = pd.merge(app, credit, on="ID")

print(data.head())
```

```python
# Convert age from days to years
data["AGE"] = (-data["DAYS_BIRTH"] / 365).astype(int)

print(data[["AGE"]].head())
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Your dataset
# app = pd.read_csv("your_dataset.csv")

# Create LabelEncoder
label_encoder = LabelEncoder()

# Select categorical columns
categorical_columns = app.select_dtypes(include=['object']).columns

# Apply Label Encoding
for column in categorical_columns:
    app[column] = label_encoder.fit_transform(app[column])

# Display encoded dataset
print(app.head())
```
