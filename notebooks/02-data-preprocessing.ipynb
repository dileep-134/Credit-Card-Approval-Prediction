```python
# Data Pre-processing Notebook
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Load data
app = pd.read_csv('../data/raw/application_record.csv')
credit = pd.read_csv('../data/raw/credit_record.csv')

# Remove duplicates
app = app.drop_duplicates()

# Handle missing values
app = app.fillna(app.mode().iloc[0])

# Save processed data
app.to_csv('../data/processed/application_processed.csv', index=False)
credit.to_csv('../data/processed/credit_processed.csv', index=False)
```
